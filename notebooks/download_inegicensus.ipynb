# Censo Poblacional INEGI

```python
from partciud.config import (api_inegi_census_2005,
api_inegi_census_2010, api_inegi_census_2020, DATA_FOLDER)
import os
import requests as r

inegi_census = [api_inegi_census_2005,
                api_inegi_census_2010,
                api_inegi_census_2020]

for file in inegi_census:
    if not os.path.exists(DATA_FOLDER / "raw" / "censo_inegi" / str(file).split("/")[-1]):
        response = r.get(file)
        response.raise_for_status()
        os.makedirs(DATA_FOLDER / "raw" / "censo_inegi", exist_ok=True)
        with open(DATA_FOLDER / "raw" / "censo_inegi" / str(file).split("/")[-1], "wb") as cfile:
            cfile.write(response.content)
```
