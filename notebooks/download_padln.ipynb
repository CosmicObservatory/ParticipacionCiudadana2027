# Lista Nominal y Padron Electoral

```python
from partciud.config import DATA_FOLDER
import datetime as dt
import requests as r
import os 

start = dt.date(2026, 1, 1)

YEARS = [2021, 2022, 2023, 2024, 2025, 2026]
start_day = dt.date(2026, 1, 1)
def create_ineurl(date):
    year = date.year
    month = date.month
    day = date.day
    urls = list()
    url1 = f"https://ine.mx/wp-content/uploads/{year}/{month:02d}/DatosAbiertos-derfe-pdln_edms_eo_{year}{month:02d}{day:02d}.xlsx"
    url2 = f"https://ine.mx/wp-content/uploads/{year}/{month:02d}/DatosAbiertos-derfe-pdln_edms_re_{year}{month:02d}{day:02d}.xlsx"
    url3 = f"https://ine.mx/wp-content/uploads/{year}/{month:02d}/DatosAbiertos-derfe-pdln_edms_sexo_{year}{month:02d}{day:02d}.xlsx"
    urls = [url1, url2, url3]
    return urls

os.makedirs(DATA_FOLDER / "raw" / "padln", exist_ok=True)

problem_dates = []
for y in YEARS:
    start_day = dt.date(y, 1, 1)
    while (dt.date(y + 1, 1, 1) - start_day).days > 0:
        files = create_ineurl(start_day)
        for v in files:
            if not os.path.exists(DATA_FOLDER / "raw" / "padln" / v.split("/")[-1]):
                try:
                    response = r.get(v)
                    response.raise_for_status()
                    with open(DATA_FOLDER / "raw" / "padln" / v.split("/")[-1], "wb") as f:
                        f.write(response.content)
                except:
                    print("Por revisar", start_day)
                    problem_dates.append(start_day)
                    break
        start_day += dt.timedelta(days=7)

```
