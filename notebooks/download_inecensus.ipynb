# Censos Electorales

```python
from partciud.config import DATA_FOLDER
from zipfile import ZipFile
from yarl import URL
import requests as r
import os

def inecensus(year):
    url = None
    if year==2024:
        url = "https://ine.mx/wp-content/uploads/2019/09/DatosAbiertos_DECEyEC_ConteosCensales2024.zip"
    else:
        url = f"https://ine.mx/wp-content/uploads/2022/04/DatosAbiertos_DECEyEC_ConteosCensales{year}.zip"
    return url

YEARS = [2000 + i for i in range(9,25)]

for y in YEARS:
    url = inecensus(y)
    try:
        if not os.path.exists(DATA_FOLDER / "raw" / "censo_electoral" / url.split("/")[-1]):
            response = r.get(url)
            response.raise_for_status()
            os.makedirs(DATA_FOLDER / "raw" / "censo_electoral", exist_ok=True)
            with open(DATA_FOLDER / "raw" / "censo_electoral" / url.split("/")[-1], "wb") as file:
                file.write(response.content)
    except:
        print("Año", y)

```

Una vez vamos 

```python
folder_icensus = os.listdir(DATA_FOLDER / "raw" / "censo_electoral")

for icfile in folder_icensus:
    if not os.path.exists(DATA_FOLDER / "interim" / "ine_censo" / icfile):
        file = DATA_FOLDER / "raw" / "censo_electoral" / icfile
        os.makedirs(DATA_FOLDER / "interim" / "ine_censo", exist_ok=True)
        with ZipFile(file, "r") as icf:
            icf.extractall(DATA_FOLDER / "interim" / "ine_censo" / icfile)
```
