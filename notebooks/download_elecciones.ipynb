# Censos Electorales

Ahora, vamos a descargar los censos electorales. Partiremos del WebScraping para obtener los archivos.

## WebScraping

```python
from partciud.config import url_elecciones
from partciud.file import filenorm
from bs4 import BeautifulSoup
import requests as r


response = r.get(url_elecciones)
response.raise_for_status()

soup = BeautifulSoup(response.content, "lxml")

file_ref = soup.find("select")
files = file_ref.find_all("option")

filename = [f"{f.text}.xlsx" for f in files if f.text != ""]

files_elecciones = filenorm(filename)
```
